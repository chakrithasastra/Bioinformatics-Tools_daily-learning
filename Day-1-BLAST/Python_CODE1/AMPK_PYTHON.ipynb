```python
from Bio.Blast import NCBIXML
import matplotlib.pyplot as plt
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
blast_record = NCBIXML.read(open("EWVFDSFN015-Alignment_AMPK.xml"))
```

```python
hit_names = []
hit_scores = []

for alignment in blast_record.alignments:
    hit_names.append(alignment.title)
    hit_scores.append(alignment.hsps[0].score)  # first HSP score
```

```python
plt.figure(figsize=(10,6))
plt.barh(hit_names, hit_scores, color='Lavender')
plt.xlabel("Alignment Score")
plt.ylabel("Hit Description")
plt.title("Top BLAST Hits for AMPK Query")
plt.gca().invert_yaxis()  # highest score at top
plt.show()
```

```python
for alignment in blast_record.alignments[:5]:
    print(f"Hit: {alignment.title}")
    print(f"Length: {alignment.length}")
    print(f"Score: {alignment.hsps[0].score}")
    print(f"E-value: {alignment.hsps[0].expect}\n")
```
